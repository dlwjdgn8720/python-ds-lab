### 4. 선택

```python
# '사용일자' 컬럼을 인덱스로 지정하여 df2 객체를 생성
df2 = df.set_index('사용일자')
df2
```

```python
# 일반 데이터 검색 : 팬시, 불린, query
# '노선명' 컬럼의 '2호선' 데이터 조회
# 불린 인덱싱 조건 : df2['노선명'] == '2호선'
df2[df2['노선명'] == '2호선']
```

```python
# 일반 데이터 검색 : 팬시, 불린, query
# '노선명' 컬럼의 '2호선' 데이터 조회
# query 조건 : '노선명 == "2호선"'
df2.query('노선명 == "2호선"')
df2.query('노선명 == "2호선"')['노선명'].count()
```

```python
# '승차총승객수'이 50000명 이상인 '역명' 조회
# 불린 인덱싱 : df2['승차총승객수'] >= 50000
df2[df2['승차총승객수'] >= 50000]['역명']
df2[df2['승차총승객수'] >= 50000]['역명'].unique()
```

```python
# '승차총승객수'이 50000명 이상인 '역명' 조회
# 쿼리 : '승차총승객수 >= 50000'
df2.query('승차총승객수 >= 50000')
df2.query('승차총승객수 >= 50000')['역명'].unique()
```

```python
# '승차총승객수'이 50000명 이상이고, '역명' 이름이 '강남'인 데이터 조회
df2[(df2['승차총승객수'] >= 50000) & (df2['역명'] == '강남')]
df2[(df2['승차총승객수'] >= 50000) & (df2['역명'] == '강남')].count()
```

```python
# '승차총승객수'이 50000명 이상이고, '역명' 이름이 '강남'인 데이터 조회
df2.query('승차총승객수 >= 50000 & 역명 == "강남"')
df2.query('승차총승객수 >= 50000 & 역명 == "강남"').count()
```

```python
# '승차총승객수' 컬럼의 최대값
df['승차총승객수'].max()
```

```python
# '승차총승객수' 컬럼의 최소값
df['승차총승객수'].min()
```

```python
# '승차총승객수' 컬럼의 최대값, '하차총승객수' 컬럼의 최소값을 조회
df.agg(
    {
        '승차총승객수': 'max',
        '하차총승객수': 'min'
    }
)
```

```python
# '승차총승객수','하차총승객수' 컬럼의 최대값, 최소값, 평균값을 조회
df[['승차총승객수', '하차총승객수']].agg(['max', 'min' ,'mean', 'median'])
```
