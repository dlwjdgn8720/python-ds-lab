```python
%matplotlib inline
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
import platform

if platform.system() == 'Windows':
    matplotlib.rc('font', family='Malgun Gothic')
elif platform.system() == 'Darwin':
    matplotlib.rc('font', family='AppleGothic')
else:
    matplotlib.rc('font', family='DejaVu Sans')

plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['figure.dpi'] = 100

print('NumPy 버전:', np.__version__)
print('환경 설정 완료')
```

```python
season = pd.Series(['봄','여름','가을','겨울','봄','봄','여름','가을'])
```

```python
# 도수분포 확인
season.value_counts()
```

```python
# 비율 게산
(season.value_counts() / season.size).round(2)
```

```python
# 시각화 - 막대그래프, 원그래프 
fd = season.value_counts()
fd.plot(
    kind = 'bar',
    figsize=(5,4),
    color=('coral')
)
plt.xlabel('계절')
plt.ylabel('빈도')
plt.title('계절별 빈도수')
plt.show()
```

```python
# 시각화 원그래프
fd = season.value_counts()
fd.plot(
    kind = 'pie',
    autopct='%1.0f%%',
)
plt.ylabel('')
plt.title('계절별 빈도수')
plt.show()
```

```python
# 시각화 - 막대 그래프, 원 그래프
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# 막대 그래프
fd.plot.bar(
    ax = axes[0],
    color=('coral'),
    title= '계절별 빈도수'
)

# 막대 그래프
fd.plot.pie(
    ax = axes[1],
    color=('coral'),
    title= '계절별 빈도수',
    ylabel=''
)

plt.show()
```

### 2. 수치형 데이터

```python
from scipy import stats

# 정상적인 몸무게 데이터
weight1 = pd.Series([60, 62, 65, 66, 64, 63])

# 이상치 데이터 추가
weight2 = pd.Series([1, 60, 62, 65, 66, 64, 63, 150])
```

```python
# 평균 구하기
print(f"mean1: {weight1.mean():.1f}")
print(f"mean2: {weight2.mean():.1f}")

# 중간값 구하기
print(f"median1: {weight1.median():.1f}")
print(f"median2: {weight2.median():.1f}")

# 절사평균 구하기
print(f"절사평균(20%): {stats.trim_mean(weight2, 0.2):.1f}")
```
