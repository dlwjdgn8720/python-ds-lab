# Index란?
- Pandas의 Index는 데이터프레임과 시리즈의 각 행(row)을 고유하게 식별하는 역할을 수행함
- 관계형 데이터베이스의 Primary key(PK)와 유사한 개념
- 특징
    - 데이터 식별 역할 수행
    - 숫자,문자열도 가능
    - 생성 후 직접 수정 불가
    - reset.index() 함수를 이용하여 재생성 가능
