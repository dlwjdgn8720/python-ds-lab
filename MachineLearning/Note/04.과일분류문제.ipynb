#### 머신러닝을 위한 데이터 준비

- Feature Data와 Target Data 분리

```python
# Feature Data
fruitsData = fruits[['length', 'weight']]
fruitsData.info()
```

```python
# Target Data
fruitsTarget = fruits['name']
fruitsTarget.info()
```

### 최근접이웃(KNN : K Nearest Neighbors) 분류 알고리즘

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Model 생성
kn = KNeighborsClassifier() # default 5, HyperParameter 조정 : n_neightbors=3
```

```python
#학습시키기
kn.fit(fruitsData, fruitsTarget)
```

```python
kn.score(fruitsData, fruitsTarget)
```

```python
# 예측해보기
# 길이가 30cm이고 무게가 600g인것은 무엇인가?
kn.predict([[30,600]])
```

```python
plt.scatter(
    apple.length,
    apple.weight
)

plt.scatter(
    berry.length,
    berry.weight
)

plt.scatter(
    30,
    600,
    c='red',
    marker = '^'
)

plt.xlabel('Length')
plt.ylabel('Weight')

plt.show()
```
