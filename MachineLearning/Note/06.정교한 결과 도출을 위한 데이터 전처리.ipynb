```python
# numpy의 column stack과 row stack
testA = np.array([1, 2, 3, 4])
testB = np.array([10, 20, 30, 40])

# 2개의 Array를 묶어서 컬럼기반의 Array로 만들기
combineA = [[i,j] for i,j in zip(testA, testB)]
print(np.array(combineA))
print("-" * 50)

#numpy에서 컬럼기반
combineB = np.column_stack((testA, testB))
print(combineB)
print("-" * 50)

#numpy에서 hstack
combineC = np.hstack((testA, testB))
print(combineC)
print("-" * 50)

#numpy에서 index 기반
combineD = np.vstack((testA, testB))
print(combineD)
```

```python
# fruits의 feature
fruits_data = np.column_stack((fruits.length, fruits.weight))
fruits_data[:5]
```

```python
fruits_data.shape
```

```python
# Target
fruits_target = np.array(fruits.name)
fruits_target[:5]
```

```python
from sklearn.model_selection import train_test_split
```

```python
train_input, test_input, train_target, test_target = \
train_test_split(
    fruits_data, 
    fruits_target,
    random_state=42,
    stratify=fruits_target # target의 비율을 검토
)
```

```python
# kNN을 이용한 예측
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Model 만들기
kn = KNeighborsClassifier() # n_neighbors=5

# 학습시키기
kn.fit(train_input, train_target)

# 검증하기
print(kn.score(train_input, train_target))
print(kn.score(test_input, test_target))
```

---
### 수상한 과일

```python
# length가 25cm이고 weight가 150g인 과일은?
kn.predict([[25, 150]])
```

### 기준을 맞추자

```python
# 이웃의 위치를 출력
plt.scatter(
    train_input[:, 0],
    train_input[:, 1]
)

plt.scatter(
    25,
    150,
    marker = '^',
    c = 'red'
)

plt.scatter(
    train_input[indexs, 0],
    train_input[indexs, 1],
    marker='D'
)

plt.xlim(0, 1000)
plt.xlabel('Length')
plt.ylabel('Weight')

plt.show()
```

> data가 Length에는 거의 영향을 받지 않고 weight에서 영향을 많이 받는다.           
중요한 Feature는 Weight

```python
# length와 weight의 평균과 표준편차
mean = np.mean(train_input, axis=0)
mean
```

```python
std = np.std(train_input, axis=0)
std
```

#### train_data를 표준점수(z점수) 변경
표준점수 = (Feature - 평균) / 표준편차

```python
train_scaled = (train_input - mean) / std
train_scaled[:5]
```

### sklearn을 이용

### 전처리 데이터로 모델 훈련하기

```python
# 예측할 데이터도 표준점수 변경이 필요
new = ([25, 150] - mean) / std
new
```

### kNN

```python
print(kn.score(train_scaled3, train_target))
print(kn.score(test_scaled, test_target))
```
