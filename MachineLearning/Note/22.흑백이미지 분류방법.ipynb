##### 군집 알고리즘
- 과일 데이터 사진을 보고 사과, 파인애플, 바나나로 3분류 하기

```python
import numpy as np
```

```python
fruits = np.load('../Data/fruits.npy')
fruits.shape
```

> 흑백 이미지
300개의 이미지가 100*100 픽셀로 구성                         
100개는 사과, 나머지 100개씩은 파인애플과 바나나로 구성

```python
fruits[0,0,:]
```

> 1장의 흑백 이미지중 1번행만 출력한 것이다.                        
최대숫자가 255이며 255는 흰색이고 0으로 근접할 수록 검은색이다.

```python
# 이미지로 확인
import matplotlib.pyplot as plt
```

```python
plt.imshow(fruits[0], cmap='gray_r')
plt.show()
```

```python
# 사과, 파인애플, 바나나 이미지를 나란히 보기
fig, axs = plt.subplots(1, 3, figsize=(20, 5))
axs[0].imshow(fruits[0], cmap='gray')
axs[1].imshow(fruits[100], cmap='gray')
axs[2].imshow(fruits[200], cmap='gray')
plt.show()
```

- 현재 3차원의 배열 분석을 위해 2차원 배열로 변경
- 배열중 page단의 변수는 apple, pineapple, banana로 분류하게 변경

```python
apple = fruits[0:100].reshape(-1, 100*100)
pineapple = fruits[100:200].reshape(-1, 100*100)
banana = fruits[200:300].reshape(-1, 100*100)
```

```python
print(apple.shape)
```

### 픽셀값 분석하기

```python
# apple의 행별 평균 구하기
apple.mean(axis=1)
```

```python
# 과일별 빈도수를 히스토그램으로 표현하기
plt.figure(figsize=(15,5))
plt.hist(np.mean(apple, axis=1))
plt.hist(np.mean(pineapple, axis=1))
plt.hist(np.mean(banana, axis=1))
plt.legend(['apple', 'pineapple', 'banana'])
plt.show()
```

```python
# Feature별(픽셀별 평균의 막대 그래프 그리기)
fig, axs = plt.subplots(1, 3, figsize=(20, 5))
axs[0].bar(range(10000), np.mean(apple, axis=0))
axs[1].bar(range(10000), np.mean(pineapple, axis=0))
axs[2].bar(range(10000), np.mean(banana, axis=0))
plt.show()
```

```python
# 평균 이미지로 그림 표현하기 : 이미지로 그리기 위해 다시 2차원 배열 변경
apple_mean = np.mean(apple, axis=0).reshape(100, 100)
pineapple_mean = np.mean(pineapple, axis=0).reshape(100, 100)
banana_mean = np.mean(banana, axis=0).reshape(100, 100)

fig, axs = plt.subplots(1, 3, figsize=(20, 5))
axs[0].imshow(apple_mean, cmap='gray')
axs[1].imshow(pineapple_mean, cmap='gray')
axs[2].imshow(banana_mean, cmap='gray')

plt.show()
```

#### 평균과 가까운 사진 고르기
- 위에서 구한 평균 이미지와 각 샘플들의 픽셀값과 비교하여 가까운 거리의              
이미지는 해당 과일일 것이다.

```python
# 사과만 잘 분류하는지 시도
abs_diff = np.abs(fruits - apple_mean)
abs_mean = np.mean(abs_diff, axis=(1, 2))
abs_mean.shape
```

```python
# 300개의 데이터중 차이가 적은 순으로 정렬하면서 해당 이미지를 불러오자
apple_index = np.argsort(abs_mean)[:100]
apple_index
```

```python
# 그림 그려보기
fig, axs = plt.subplots(10, 10, figsize=(10, 10))

for i in range(10):
    for j in range(10):
        axs[i,j].imshow(fruits[apple_index[i*10+j]], cmap='gray')
        axs[i,j].axis('off')

plt.show()
```

----
#### 바나나로 확인해 보기

```python
# 바나나만 잘 분류하는지 시도
banana_diff = np.abs(fruits - banana_mean)
abs_mean = np.mean(banana_diff, axis=(1, 2))
abs_mean.shape
```

```python
# 300개의 데이터중 차이가 적은 순으로 정렬하면서 해당 이미지를 불러오자
banana_index = np.argsort(abs_mean)[:100]
banana_index
```

```python
# 그림 그려보기
fig, axs = plt.subplots(10, 10, figsize=(10, 10))

for i in range(10):
    for j in range(10):
        axs[i,j].imshow(fruits[banana_index[i*10+j]], cmap='gray')
        axs[i,j].axis('off')

plt.show()
```
