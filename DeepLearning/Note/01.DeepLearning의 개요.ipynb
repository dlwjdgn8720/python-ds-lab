#### Optimizer
- Batch 경사 하강법
- SGD(Stochastic Gradient Descent)

#### 미니배치(mini batch)
- 배치 경사법은 너무 느리고, 리소스도 많이 사용하고, SGD는 확실히 빨리 학습하지만                      
모델이 최적화 안되는 경우가 많습니다.
- 두 방법의 절충안이 미니배치
- 전체 데이터를 계산해서 매개변수를 변경하는 대신 정해진 양만큼 계산해서              
매개변수를 최적화 하는 방법 
- 최근 딥러닝 학습에서는 대다수가 미니배치를 사용

#### 학습률(Learning Rate)
- Adagrad
- Adam


#### 딥러닝의 과대적합(Overfitting)
- 과대적합은 학습시와 검증시의 예측력의 차이가 심한 현상
- 딥러닝 모델은 학습시 매개변수가 상당히 많다는 점과 학습 횟수에 제한이                        
없으며 이 두 특징은 과대적합을 불러 올수 있다.

#### 드룹아웃(Drop out)
- 드룹아웃은 매개변수중 일정량을 학습 중간마다 무작위로 사용하지 않는 방법

#### 조기종료(Early Stopping)
- 무조건 학습 반복 횟수(Epoch)를 높이면 학습시간이 길어져서 학습데이터만                     
성능이 좋은 현상이 발생
- 어느정도 학습률이 좋으면 사용자가 정한 반복횟수 전에 작업을 종료한다.
