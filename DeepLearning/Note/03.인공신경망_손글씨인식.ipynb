```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping
```

```python
mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
x_train, x_test = x_train / 255.0, x_test / 255.0
```

```python
model = models.Sequential([
    # 28x28 2차원 이미지를 784개의 1차원 배열로 펼칩니다.
    layers.Flatten(input_shape=(28, 28)), 
    
    # 은닉층(Hidden Layer): 128개의 노드를 두고, 활성화 함수로 ReLU를 사용합니다.
    layers.Dense(128, activation='relu'), 
    
    # 과적합을 방지하기 위해 학습 시 20%의 뉴런을 무작위로 끕니다.
    layers.Dropout(0.2), 
    
    # 출력층(Output Layer): 0부터 9까지 총 10개의 숫자로 분류합니다.
    layers.Dense(10, activation='softmax') 
])
```

```python
# 모델의 컴파일 단계 (손실함수, 최적화 도구, 평가 기준 설정)
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# 검증 데이터의 손실(val_loss)을 감시합니다.
early_stopping = EarlyStopping(
    monitor='val_loss', 
    patience=3,          # val_loss가 3번의 에포크 동안 개선되지 않으면 학습을 멈춥니다.
    restore_best_weights=True  # 학습을 멈춘 후, 가장 성적이 좋았던 에포크의 가중치로 되돌립니다.
)
```

```python
print("--- 학습 시작 ---")
history = model.fit(
    x_train, y_train, 
    epochs=100,              # 최대 에포크는 100으로 넉넉하게 잡습니다.
    validation_split=0.2,    # 6만 장 중 20%(12,000장)를 검증용 데이터로 떼어놓고 감시합니다.
    callbacks=[early_stopping] # 조기 종료 기능을 적용합니다.
)
```

```python
print("\n--- 최종 테스트 데이터 평가 ---")
test_loss, test_acc = model.evaluate(x_test,  y_test, verbose=2)
print(f"\n테스트 정확도: {test_acc * 100:.2f}%")
```
